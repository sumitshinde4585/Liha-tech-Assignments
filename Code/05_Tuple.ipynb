```python
t1=(1,2,3)
t2=(4,5,6)
t3=()
for a,b in zip(t1,t2):
    t3=t3+(a+b,)
    
print(t3)
print(tuple(x+y for x,y in zip(t1,t2)))
```

```python
t= (1, 2, 2, 3, 4, 4, 4)
ans=()
for i in t:
    if t.count(i) >1 and i not in ans:
        ans+=(i,)
print(ans)

print(tuple(i for i in set(t) if t.count(i)>1))
```

```python
#4.  Write a Python program to sort a list of tuples by the second element of each tuple.


data= [('A', 75), ('B', 92), ('C', 80)]
for i in range(len(data)):
    for j in range(i+1,len(data)):
        if data[i][1]>data[j][1]:
            data[i],data[j]=data[j],data[i]
print(data)

data= [('A', 75), ('B', 92), ('C', 80)]

data.sort(key=lambda x:x[1])

print(data)
```

```python
# 7.Write a Python program to calculate column-wise sums of a tuple-based matrix.
t= ((1, 2, 3), (4, 5, 6), (7, 8, 9))
ans=()
for a,b,c in zip(*t):
    ans+=(a+b+c,)
print(ans)

print(tuple(sum(col) for col in zip(*t)))
```

```python
#9. Write a Python program to interleave two tuples of equal length.
t1 = (1, 2, 3)
t2 = (10, 20, 30)

result = tuple(x for pair in zip(t1, t2) for x in pair)

print(result)
```
