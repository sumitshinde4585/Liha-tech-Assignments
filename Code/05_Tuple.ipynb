```python
#5.  Write a Python program to recursively flatten a nested tuple into a single tuple.

def flatten(t):
    ans=()
    for i in t:
        if isinstance(i,tuple):
            ans=ans+flatten(i)
        else:
            ans=ans+(i,)
    return ans
t=(1, (2, (3, 4)), 5)
print(flatten(t))


def flat(t):
    return sum((flat(i) if isinstance(i,tuple) else (i,) for i in t),())
print(flat(t))
```

```python
records = (('Amit', 'DS'), ('Neha', 'AI'), ('Ravi', 'DS'))

groups = {}

for name, dept in records:
    if dept not in groups:
        groups[dept] = ()
    
    groups[dept] = groups[dept] + (name,)

print(groups)
```

```python
from collections import defaultdict

records = (('Amit', 'DS'), ('Neha', 'AI'), ('Ravi', 'DS'))

groups = defaultdict(tuple)


for name, dept in records:
    groups[dept] = groups[dept] + (name,)

print(dict(groups))
print(list(groups))
```

```python
#9.  Write a Python program to find the longest increasing contiguous subsequence in a tuple.
t= (1, 2, 5, 3, 4, 6, 7, 2)

t = (1, 2, 3, 2, 4, 5, 6, 1)

longest = ()
current = (t[0],)

for i in range(1, len(t)):
    if t[i] > t[i - 1]:
        current = current + (t[i],)
    else:
        if len(current) > len(longest):
            longest = current
        current = (t[i],)

if len(current) > len(longest):
    longest = current

print("Longest increasing subsequence =", longest)
```

```python
t = (1, 2, 3, 2, 4, 5, 6, 1)

parts = []
start = 0

for i in range(1, len(t)):
    if t[i] <= t[i - 1]:
        parts.append(t[start:i])
        start = i

parts.append(t[start:])

result = max(parts, key=len)

print("Longest increasing subsequence =", result)
```

```python
# 10.  Write a Python program to merge two sorted tuples into one sorted tuple without using sort().

t1 = (1, 4, 7)
t2 = (2, 3, 8, 9)
i=0
j=0

res=()
while i<len(t1) and j<len(t2):
    if t1[i]<t2[j]:
        res+=(t1[i],)
        i+=1
    else:
        res+=(t2[j],)
        j+=1
while i<len(t1):
    res+=(t1[i],)
    i+=1
while j<len(t2):
    res+=(t2[j],)
    j+=1
    

print(res)
```
