```python
# 2.  Write a Python function using *args to return the sum of any number of arguments.
values = (10, 20, 30, 40, 50)
def sumNums(*args):
    s=0
    for i in args:
        s += i
    print("Sum =",s)
sumNums(*values)
```


| Code                | Meaning                                |
| ------------------- | -------------------------------------- |
| `values`            | Tuple                                  |
| `*values`           | Unpack tuple                           |
| `sumNums(10,20,30)` | 3 separate arguments                   |
| `sumNums(*values)`  | Tuple unpacked into separate arguments |
| `sumNums(values)`   | Tuple passed as 1 argument             |



sumNums((10, 20, 30, 40, 50))

then the tuple is treated as one argument, so your loop will try to do:

s += (10, 20, 30, 40, 50)

```python
# 6.  Write a recursive Python function to find the sum of digits of a number.
number= 5832
def recSum(n):
    if n==0:
        return 0
    return n%10 +recSum(n//10)    
print(recSum(number))


# 2nd apporach
def sum_digits(n):
    if n < 10:
        return n
    return int(str(n)[0]) + sum_digits(int(str(n)[1:]))

print(sum_digits(12345))
    
```

```python
#7.  Write a recursive Python function to return the nth Fibonacci number 
def fibonacci(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)
print(fibonacci(8))
```

```python
# 8.  Write a Python function to calculate income tax using slabs: 0% up to 300000, 5% for the next 300000,
# and 10% for the remaining amount

def  calTax(amount):
    tax=0
    if amount <=300000:
        tax=0
    elif 300000<amount<=600000:
        tax=(amount-300000)*0.05
    else:
        tax=(300000)*0.05 + (amount-600000)*0.10
    return tax
print(calTax(800000))


```

```python
# 10.  Write a Python function to validate a password and return both status and reason. Require at least 8characters, one uppercase letter, and one digit.
def validate_password(password):
    if len(password) < 8:
        return False, "Password must have at least 8 characters"

    if not any(ch.isupper() for ch in password):
        return False, "Password must contain one uppercase letter"

    if not any(ch.isdigit() for ch in password):
        return False, "Password must contain one digit"

    return True, "Password is valid"


print(validate_password("hello"))
print(validate_password("Hello123"))
```
