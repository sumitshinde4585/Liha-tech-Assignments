```python
# Write a Python function to convert Celsius temperature to Fahrenheit.
def celsius_To_fahrenheit(celsius):
    fahrenheit =(celsius*9/5)+32
    return fahrenheit
print("25 celsius =",celsius_To_fahrenheit(25),"fahreheit")
```

```python
text ="Programming"
def countVowels(s):
    vowel =0
    for ch in s:
        if ch in "AEIOUaeiou":
            vowel+=1
    return vowel
print("Total Vowel in",text,"is",countVowels(text))
    
```

```python
def rev(s):
    return s[::-1]
print("Reverse of Python is",rev("Python"))

def rev1(s):
    for i in range(len(s)-1,-1,-1):
        print(s[i],end="")
rev1("Python")
```

```python
def fact(n):
    if n==1 or n==0:
        return 1
    return n*fact(n-1)
print(fact(5))
```

```python
numbers = [10, 20, 30, 40]
def nSum(numbers):
    t=0
    for i in range(len(numbers)):
       t+=numbers[i]
    return t
print("sum of all numbres is",nSum(numbers)) 
```

```python
numbers = [10, 20, 30, 40]
def avgNums(numbers):
    t=0
    for i in range(len(numbers)):
       t+=numbers[i]
    return t/len(numbers)
print("Avg of all numbres is",avgNums(numbers))
```

```python
# 13.  Write a Python function to check whether a string is a palindrome
def checkPalindrome(s):
        return s ==s[::-1]
print(checkPalindrome("madam"))
```

```python
principal = 10000
rate = 8
time = 2
def simpleInterest(p,r,t):
    return p*(r/100)*t
print(simpleInterest(principal,rate,time))
```
