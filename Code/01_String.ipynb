```python
s="LihaTech@2026"
u=0
l=0
d=0
sp=0
for i in s:
    if i.isupper():
        u+=1
    elif i.islower():
        l+=1
    elif i.isdigit():
        d+=1
    else:
        sp+=1
print("Uppercase =",u,"  Lowercase =",l," Digits =",d," Special =",sp)
```

### SECTION B - INTERMEDIATE



```python
s = "Python is fun and python is powerful"
d={}
l=s.split()
for i in l:
    if i.lower() not in d:
        d[i.lower()]=1
    else:
        d[i.lower()]+=1
for k,v in d.items():
    print(k,":",v)
```

```python
s="Python makes programming enjoyable"
l=s.split()
length=0
lw=""
for i in l:
    if len(i)>length:
        lw=i
        length=len(i)
print("Longest word =",lw)
print("Length =",length)
```

```python
s="aaaaa"
sub="aa"
p=[]
for i in range(len(s)-len(sub)+1):
    if s[i:i+len(sub)] ==sub :
        p.append(i)
print(p)
```

```python
s="aaabbccccd"
count = 1
result = ""
for i in range(len(s)):
    if i< len(s)-1 and s[i] == s[i+1]:
        count+=1
    else:
        result+=s[i]+str(count)
        count=1
print(result)
```

```python
s="The quick brown fox jumps over the lazy dog"
s=s.lower()
letters=set()
for ch in s:
    if ch.isalpha():
        letters.add(ch)
print("Pengram =",len(letters)==26)
```
