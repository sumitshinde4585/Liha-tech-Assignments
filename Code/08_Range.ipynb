```python
n=int(input("Enter n for sum of even numbers:"))
sum=0
for i in range(2,n+1,2):
    sum+=i
print(sum)
```

```python
n=int(input("Enter n for all divisors:"))
for i in range(1,n+1):
    if n%i==0:
        print(i,end=" ")
```

```python
start=int(input("Enter Strat:"))
end=int(input("Enter End :"))
for i in range(start,end+1):
    if i >1:
        for j in range(2,(i//2)+1):
            if i%j==0:
                break
        else:
            print(i,end=" ")
```
