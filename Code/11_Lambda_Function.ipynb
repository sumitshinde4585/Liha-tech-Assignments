| Function   | Main purpose                     |
| ---------- | -------------------------------- |
| `lambda`   | Create a small function          |
| `map()`    | Transform every item             |
| `filter()` | Select matching items            |
| `sorted()` | Sort using a custom rule         |
| `max()`    | Find maximum using a custom rule |
| `min()`    | Find minimum using a custom rule |
| `reduce()` | Combine all items into one value |

```python
# 12.  Write a Python program to sort words by their length using a lambda function.
words = ['python', 'AI', 'data', 'cloud']

result = sorted(words , key= lambda x :len(x))
print(result)
```

```python
# 13.  Write a Python program to square all numbers in a list using map() and lambda
numbers =[1,2,3,4,5,6]
ans =list(map(lambda x :x*x ,numbers))
print(ans)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
ans =list(filter(lambda x : x%2==0 ,numbers))
print(ans)
```

```python
# start with A
words = ['Apple', 'Python', 'AI', 'Data', 'Agent']
ans =list(filter( lambda x : x.startswith("A") ,words))
print(ans)
```

```python
words = ['Apple', 'Python', 'AI', 'Data', 'Agent']
ans =list(filter( lambda x : x.endswith("t") ,words))
print(ans)
```

```python
# 16.  Write a Python program to convert Celsius values to Fahrenheit using map() and lambda.
celsius = [0, 20, 30, 40]
celsius_to_fahrenheit = list(map(lambda c :(c* 9/5)+32 ,celsius))
print(celsius_to_fahrenheit)
```
