| Function   | Main purpose                     |
| ---------- | -------------------------------- |
| `lambda`   | Create a small function          |
| `map()`    | Transform every item             |
| `filter()` | Select matching items            |
| `sorted()` | Sort using a custom rule         |
| `max()`    | Find maximum using a custom rule |
| `min()`    | Find minimum using a custom rule |
| `reduce()` | Combine all items into one value |

```python
versions = [('app', '1.2.10'), ('app', '1.10.2'), ('app', '1.2.3')]

ans = sorted(versions, key=lambda x: tuple(map(int, x[1].split('.'))))

print(ans)
```

```python

points = [(3, 4), (1, 1), (-2, 2), (0, 5)]

ans = sorted(points, key=lambda x: abs(x[0]) + abs(x[1]))

print(ans)
```

```python
dates = ['2026-12-01', '2025-05-10', '2026-01-15']
ans = sorted(dates , key=lambda x: tuple(map(int ,x.split("-"))))
print(ans)
```

```python
words = ['AI', 'Python', 'MachineLearning', 'Data']
ans =max(words ,key=lambda x :len(x))
print(ans)
```

```python

calculator = {
    'add': lambda a, b: a + b,
    'subtract': lambda a, b: a - b,
    'multiply': lambda a, b: a * b,
    'divide': lambda a, b: a / b
}

a = 10
b = 5

print("Addition:", calculator['add'](a, b))
print("Subtraction:", calculator['subtract'](a, b))
print("Multiplication:", calculator['multiply'](a, b))
print("Division:", calculator['divide'](a, b))
```

```python
# 10.  Write a Python program using reduce() and lambda to find the largest number without using max().
numbers = [12, 45, 7, 91, 33]
from functools import reduce
ans = reduce(lambda x, y : x if x>y else y,numbers)
print(ans)
```
