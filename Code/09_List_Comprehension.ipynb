```python
# 2.Write a Python program to transpose a matrix using nested list comprehension.
matrix = [[1, 2, 3], [4, 5, 6]]
transpose =[[matrix[i][j] for i in range(len(matrix))] for j in range(len(matrix[0]))]
print(transpose)

# using range
print("transpose using range fun")
t=[]
for j in range(len(matrix[0])):
    r=[]
    for i in range(len(matrix)):
        r.append(matrix[i][j])
    t.append(r)
print(t)
```

```python
# 2.  Write a Python program to generate all pairs whose sum is equal to a given target.
numbers = [1, 2, 3, 4, 5, 6]
t=7
pairs=[(numbers[i],numbers[j]) for i in range(len(numbers)) for j in range(i+1,len(numbers)) if numbers[i]+numbers[j]==t]
print(pairs)
```

```python
# 3.  Write a Python program to create the transpose of a 3 x 3 matrix using nested list comprehension.
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
transpose = [[matrix[i][j] for i in range(len(matrix))] for j in range(len(matrix[0]))]
print(transpose)
```

```python
# 4.  Write a Python program to flatten a three-level nested list into a single list.
data = [[[1, 2], [3]], [[4, 5], [6]]]
flatten= [x for a in data for b in a for x in b]
print(flatten)
```

```python
# 5.Write a Python program to calculate the sum of each row of a matrix using list comprehension.
matrix = [[2, 4, 6], [1, 3, 5], [10, 20, 30]]
sumOfRow=[sum(x) for x in matrix ]
print(sumOfRow)
```

```python
# 6.  Write a Python program to assign grades to marks using a conditional list comprehension: A for 90+, B
# for 75-89, C for 60-74, and D below 60
marks = [95, 82, 67, 54, 90]
grades = ['A' if x >=90 else 'B' if x >=75 else 'C' if x >=60 else 'D' for x in marks]
print(grades)
```

```python
# 7.  Write a Python program to convert comma-separated student records into tuples
records = ['Asha,85', 'Ravi,72', 'Neha,91']
ans = [(name ,int(marks)) for name , marks in (record.split(",") for record in records)]
print(ans)
```

```python
# 5.Write a Python program to create a multiplication table from 1 to 5 as a nested list.
ans = [[i*j for j in range(1,6)] for i in range(1,6)]
print(ans)
```

```python
# 9.  Write a Python program to generate all unique Pythagorean triples where 1 <= a < b < c <= 20.
triples = [(a ,b ,c) for a in range(1,21) for b in range(a,21) for c in range(b,21) if a*a+b*b==c*c]
print(triples)
```

```python
# 10.  Write a Python program to generate sliding windows of size 3 from a list using list comprehension.
numbers = [10, 20, 30, 40, 50, 60]
window = [ numbers[i:i+3] for i in range(len(numbers)-2) ]
print(window)
```
