```python
# 1.Write a Python program to flatten a nested list of any depth into a single list.

l=[1, [2, [3, 4], 5], 6]
res=[]
def flatten(l):
    for item in l:
        if isinstance(item,list):
            flatten(item)
        else:
            res.append(item)

flatten(l)
print(res)
                
    
```

```python
# 2. Write a Python program to find the contiguous subarray having the maximum sum

l=[-2, 1, -3, 4, -1, 2, 1, -5, 4]
current=l[0]
max_sum=l[0]

start=0
end=0
temp_start=0

for i in range(1,len(l)):
    if l[i]>current+l[i]:
        current=l[i]
        temp_start=i
    else:
        current=current+l[i]
    
    if current>max_sum:
        max_sum=current
        start=temp_start
        end=i
        
print("Maximun subarray =" ,l[start:end+1])
print("Maximum sum =", max_sum)

```

```python
# 3. Write a Python program to build a list where each element is the product of all other elements, without using division

# time complexity  O(n^2)
nums=[1,2,3,4]
res=[]
for i in range(len(nums)):
    product=1
    for j in range(len(nums)):
        if i!=j:
            product*=nums[j]
    res.append(product)
print(res)


# time complexity  O(n)
nums=[1,2,3,4]
result=[1]*len(nums)
prefix=1
for i in range(len(nums)):
    result[i]=prefix
    prefix*=nums[i]
    
suffix=1
for j in range(len(nums)-1,-1,-1):
    result[j]*=suffix
    suffix*=nums[j]
    
print(result)   
```

```python
# 5. Write a Python program to merge overlapping intervals represented as a list of [start, end] pairs.

l=[[1, 3], [2, 6], [8, 10], [9, 12]]
l.sort()

result=[]
for item in l:
    if not result or result[-1][1]<item[0]:
        result.append(item)
    else:
        result[-1][1]=max(result[-1][1],item[1])
print(result)
```

```python
# 6. Write a Python program to rotate a square matrix represented as a nested list by 90 degrees clockwise

matrix=[[1, 2, 3], [4, 5, 6], [7, 8, 9]]
n=len(matrix)
result=[[0]*n for _ in range(n)]

for i in range(n):
    for j in range(n):
        result[j][n-1-i]=matrix[i][j]
print(result)

# 90° Clockwise = Reverse rows → Transpose
print([list(row) for row in zip(*matrix[::-1])])

# Another common method: Transpose + Reverse
res = [list(row) for row in zip(*matrix)]
for row in res:
    row.reverse()
print(res)
```

```python
# 7. Write a Python program to print the elements of a matrix in spiral order

l=[[1,2,3],[4,5,6],[7,8,9]]
res=[]

top=0
bottom=len(l)-1
left=0
right=len(l[0])-1

while top <= bottom and left <= right:
    # left->right
    for i in range(left,right+1):
        res.append(l[top][i])
    top+=1
    
    #  top->bottom
    for i in range(top,bottom+1):
        res.append(l[i][right])
    right-=1
    
    # right->left
    for i in range(right,left-1,-1):
        res.append(l[bottom][i])
    bottom-=1
        
    # bottom->top
    for i in range(bottom,top-1,-1):
        res.append(l[i][left])
    left+=1
        
print(res)
```

```python
# 8. Write a Python program to find the longest sequence of consecutive integers in an unsorted list.

numbers = [100, 4, 200, 1, 3, 2]
num_set=set(numbers)
longest=[]

for num in num_set:
    
    if num -1 not in num_set:
        current=num
        sequence=[]
        
        while current in num_set:
            sequence.append(current)
            current+=1
        
        if len(sequence) >len(longest):
            longest=sequence
        
print("Longest sequense =",longest)
print("Length =",len(longest))
```

```python
# 9. Write a Python program to merge two already sorted lists into one sorted list without calling sort()

l1 = [1, 3, 5, 7]
l2 = [2, 4, 6, 8]
i=0
j=0
ans=[]
while i<len(l1) and j<len(l2):
    
    if l1[i]<=l2[j]:
        ans.append(l1[i])
        i+=1
    else:
        ans.append(l2[j])
        j+=1
    
while i<len(l1):
    ans.append(l1[i])
    i+=1
while j<len(l2):
    ans.append(l2[j])
    j+=1
print(ans)
```
