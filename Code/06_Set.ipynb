1. set → mutable ✅

2. frozenset → immutable 🔒

3. frozenset supports operations like union, intersection, difference, etc.

4. But methods like add(), remove(), and update() are not available.

```python
fs=frozenset([1,2,3,4,4,5,5,6])
print(fs)  # frozenset({1, 2, 3, 4, 5, 6})

s2=frozenset([90,88,77,5,1])
print(s2)

## Set operations are allowed
print("Union:", fs | s2)
print("Intersection:", fs & s2)

#  s2.add(55)   #  Error: frozenset has no add() method
```

## SECTION C - ADVANCED

```python
nums={100, 4, 200, 1, 3, 2}
longest=[]
for num in nums:
    if num-1 not in nums:
        current=num
        seq=[num]
        
        while current+1 in nums:
            current+=1
            seq.append(current)
        if len(seq)>len(longest):
            longest=seq
        
print("longest consecutive sequence length :", len(longest))
print(*longest)
```

```python
nums = [2, 4, 3, 5, 7, 8, 1, 6, 4]
target = 9

seen=set()
pairs=set()

for num in nums:
    complement= target-num
    
    if complement in seen:
        pairs.add(tuple(sorted((num,complement))))
    seen.add(num)
print("Unique pairs",pairs)
```
