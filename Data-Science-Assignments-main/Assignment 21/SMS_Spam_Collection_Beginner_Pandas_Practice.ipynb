## About the Dataset

The **spam.csv** dataset contains text messages (SMS) that are labelled as either `ham` (normal message) or `spam`.

**Source:** Kaggle - SMS Spam Collection Dataset (https://www.kaggle.com/datasets/uciml/sms-spam-collection-dataset).

The dataset contains:

- **5,572 rows**
- **5 columns (2 useful columns and 3 almost empty unnamed columns)**
- **The three unnamed columns contain mostly missing values**; you will remove them in Section 2

### Columns in the Dataset

| Column | Description |
|---|---|
| `v1` | Label of the message (`ham` or `spam`) |
| `v2` | Text of the message |
| `Unnamed: 2`, `Unnamed: 3`, `Unnamed: 4` | Almost empty extra columns created by the file format |
