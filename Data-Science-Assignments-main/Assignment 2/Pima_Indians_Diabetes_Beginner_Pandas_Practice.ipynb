## About the Dataset

The **PimaIndiansDiabetes.csv** dataset contains health-related measurements for **768 records**.

It contains **9 columns**:

| Column | Description |
|---|---|
| `TimesPregnant` | Number of times pregnant |
| `GlucoseConcentration` | Glucose concentration measurement |
| `BloodPrs` | Blood pressure measurement |
| `SkinThickness` | Skin thickness measurement |
| `Serum` | Serum-related measurement |
| `BMI` | Body Mass Index |
| `DiabetesFunct` | Diabetes function value |
| `Age` | Age of the person |
| `Class` | Diabetes outcome class represented by 0 or 1 |

### Important Observation

The dataset does not contain standard `NaN` missing values.

However, some health-related columns may contain **0 values**. For this beginner exercise, we will treat the dataset as provided and focus mainly on learning Pandas operations.
