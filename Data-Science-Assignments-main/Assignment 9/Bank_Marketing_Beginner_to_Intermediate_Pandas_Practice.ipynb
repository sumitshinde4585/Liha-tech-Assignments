# Exploratory Data Analysis and Insights on Bank Marketing Using Pandas

### Beginner-to-Intermediate Pandas Practice Notebook

This notebook is designed for students who are learning how to explore customer and campaign data using **Pandas**.

The exercises begin with basic DataFrame operations and gradually introduce:

- Loading and inspecting a CSV file
- Checking data quality
- Exploring categorical customer information
- Selecting, filtering, and sorting records
- Grouping and aggregating campaign results
- Creating derived columns
- Examining correlations
- Creating basic visualizations
- Exporting an analyzed DataFrame


## About the Dataset

The **BankMarketing.csv** dataset contains **4,521 customer-contact records** from direct-marketing campaigns conducted by a Portuguese banking institution.

It contains **17 columns**:

| Group | Example columns | Description |
|---|---|---|
| Customer profile | `age`, `job`, `marital`, `education` | Basic customer characteristics |
| Financial information | `credit_default`, `balance`, `housing_loan`, `personal_loan` | Account balance and loan information |
| Current contact | `contact_type`, `contact_day`, `contact_month`, `duration_seconds` | Information about the latest campaign contact |
| Campaign history | `campaign_contacts`, `days_since_previous_contact`, `previous_contacts`, `previous_outcome` | Current and previous contact history |
| Outcome | `subscribed` | Whether the customer subscribed to a term deposit (`yes` or `no`) |

### Important Observation

This assignment uses the smaller 10% sample supplied by the repository. It is suitable for classroom practice and remains small enough for a Git repository.

Some categorical columns contain the text value `unknown`. This is a recorded category, not a standard Pandas missing value. The value `-1` in `days_since_previous_contact` indicates that the customer was not previously contacted.

The original semicolon-separated file was converted to a standard comma-separated CSV, and abbreviated column names were replaced with descriptive names. The records were not otherwise changed.
