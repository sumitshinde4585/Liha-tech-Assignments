## About the Dataset

The **WineQualityRed.csv** dataset contains laboratory measurements and quality ratings for **1,599 red vinho verde wine samples** from Portugal.

It contains **12 columns**:

| Column | Description |
|---|---|
| `fixed_acidity` | Acids that do not evaporate easily |
| `volatile_acidity` | Volatile acids that can affect taste at high levels |
| `citric_acid` | Citric acid concentration |
| `residual_sugar` | Sugar remaining after fermentation |
| `chlorides` | Salt concentration |
| `free_sulfur_dioxide` | Free sulfur dioxide concentration |
| `total_sulfur_dioxide` | Total sulfur dioxide concentration |
| `density` | Wine density |
| `pH` | Acidity level measured on the pH scale |
| `sulphates` | Sulphate concentration |
| `alcohol` | Alcohol percentage by volume |
| `quality` | Sensory quality score |

### Important Observation

The dataset contains no standard missing values, but it does contain duplicate rows. This assignment uses those duplicates to practice data-quality checks and cleaning.

The original file uses semicolons as separators. It has been converted to a standard comma-separated CSV, and spaces in column names have been replaced with underscores.

**Source:** [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/186/wine+quality)
