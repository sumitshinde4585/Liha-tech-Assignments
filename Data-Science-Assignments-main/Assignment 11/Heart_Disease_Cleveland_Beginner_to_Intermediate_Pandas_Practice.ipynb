# Exploratory Data Analysis and Insights on the Cleveland Heart Disease Dataset Using Pandas

## About the Dataset

The **HeartDiseaseCleveland.csv** dataset contains **303 patient records** from the Cleveland database included in the UCI Heart Disease collection.

It contains **14 columns**:

| Column | Description |
|---|---|
| `age` | Age in years |
| `sex` | Encoded sex: 0 female, 1 male |
| `chest_pain_type` | Encoded chest-pain category from 1 to 4 |
| `resting_blood_pressure` | Resting blood pressure in mm Hg |
| `cholesterol` | Serum cholesterol in mg/dl |
| `fasting_blood_sugar` | Whether fasting blood sugar exceeded the dataset threshold |
| `resting_ecg` | Encoded resting electrocardiogram result |
| `max_heart_rate` | Maximum heart rate achieved |
| `exercise_induced_angina` | Encoded exercise-induced angina indicator |
| `st_depression` | ST depression induced by exercise relative to rest |
| `st_slope` | Encoded slope of the peak exercise ST segment |
| `num_major_vessels` | Number of major vessels colored by fluoroscopy |
| `thalassemia` | Encoded thalassemia-test result |
| `diagnosis` | Original outcome: 0 indicates absence and 1–4 indicate presence of heart disease |

### Important Observation

The original processed file does not include a header and uses a question mark (`?`) for missing information. A header has been added, but the record values and missing-value markers are unchanged. Use `na_values='?'` when loading the CSV.

This notebook is for data-analysis practice only. It must not be used to diagnose patients or make medical decisions.
