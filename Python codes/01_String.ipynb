```python
# 2 Write a Python program to find the longest palindromic substring in a given string
def longest_palindrome(s):
    if not s:
        return ""

    start, end = 0, 0

    def expand(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right - 1   # return the actual palindrome bounds

    for i in range(len(s)):
        l1, r1 = expand(i, i)         # odd length (center = single char)
        l2, r2 = expand(i, i + 1)     # even length (center = between two chars)

        if r1 - l1 > end - start:
            start, end = l1, r1
        if r2 - l2 > end - start:
            start, end = l2, r2

    return s[start:end + 1]


s = "babad"
print(longest_palindrome(s))
```

```python
# 3 Write a Python program to encrypt a string using a Caesar cipher with a given shift.

s = "Hello World"
shift = 3
result=""

for ch in s:
    if ch.isalpha():
        if ch.isupper():
            result += chr((ord(ch)-ord("A")+shift)%26+ord("A"))
        else:
            result += chr((ord(ch)-ord("a")+shift)%26+ord("a"))
    else:
        result += ch
print(result)
```

```python
# 6.Write a Python program to find the longest common prefix among a list of strings.

l=['flower', 'flow', 'flight']
prefix = l[0]
for word in l[1:]:
    while not word.startswith(prefix):
        prefix = prefix[:-1]
print(prefix)
```

```python
# 7.Write a Python program to group a list of words into anagram groups.

l=['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
d={}
ans=[]
for i in l:
    if "".join(sorted(i)) not in d:
        d["".join(sorted(i))] =[i]
    else:
        d["".join(sorted(i))].append(i)
for v in d.values():
    ans.append(v)
print(ans)
```

```python
# 9.Write a Python program to determine whether a string is made by repeating a smaller substring.

s="abcabcabcabc"
for i in range(1,len(s)):
    if len(s)% i == 0:
        sub = s[:i]
        count =len(s)//i
        
        if sub*count == s:
            print("Repeated pattern =", sub)
            print("Repeat count =", count)
            break
```

```python
# 10.Write a Python program to find the smallest substring that contains all characters of a given pattern.

s = "ADOBECODEBANC"
pattern = "ABC"

count = {}
for ch in pattern:
    count[ch] = count.get(ch, 0) + 1

left = 0
formed = 0
required = len(pattern)
window = {}

min_len = float("inf")
start = 0

for right in range(len(s)):
    ch = s[right]
    window[ch] = window.get(ch, 0) + 1

    if ch in count and window[ch] == count[ch]:
        formed += 1

    while formed == required:
        if right - left + 1 < min_len:
            min_len = right - left + 1
            start = left

        left_ch = s[left]
        window[left_ch] -= 1

        if left_ch in count and window[left_ch] < count[left_ch]:
            formed -= 1

        left += 1

print("Minimum window =", s[start:start + min_len])
```
