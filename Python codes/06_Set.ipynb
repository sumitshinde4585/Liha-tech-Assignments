2. frozenset → immutable 🔒

3. frozenset supports operations like union, intersection, difference, etc.

4. But methods like add(), remove(), and update() are not available.

```python
fs=frozenset([1,2,3,4,4,5,5,6])
print(fs)  # frozenset({1, 2, 3, 4, 5, 6})

s2=frozenset([90,88,77,5,1])
print(s2)

## Set operations are allowed
print("Union:", fs | s2)
print("Intersection:", fs & s2)

#  s2.add(55)   #  Error: frozenset has no add() method
```

```python
# 1.  Write a Python program to find the longest sequence of consecutive integers using a set for fast lookup.
nums={100, 4, 200, 1, 3, 2}
longest=[]
for num in nums:
    if num-1 not in nums:
        current=num
        seq=[num]
        
        while current+1 in nums:
            current+=1
            seq.append(current)
        if len(seq)>len(longest):
            longest=seq
        
print("longest consecutive sequence length :", len(longest))
print(*longest)
```

```python
# 2.  Write a Python program to find all unique pairs whose sum equals a target using a set
nums = [2, 4, 3, 5, 7, 8, 1, 6, 4]
target = 9

seen=set()
pairs=set()

for num in nums:
    complement= target-num
    
    if complement in seen:
        pairs.add(tuple(sorted((num,complement))))
    seen.add(num)
print("Unique pairs",pairs)
```

```python
#3.  Write a Python program to generate the power set of a set.

s={1,2,3}
power_set=[set()]

for ele in s:
    new_sets=[]
    
    for subset in power_set:
        new_sets.append(subset | {ele})
        
    power_set.extend(new_sets)
print(power_set)
```

```python
#4. Write a Python program to validate that a Sudoku row contains digits 1 to 9 exactly once using a set.
row = [5, 3, 4, 6, 7, 8, 9, 1, 2]

if len(row) == 9 and set(row) == set(range(1, 10)):
    print("Valid Sudoku row")
else:
    print("Invalid Sudoku row")
```

```python
#5. Write a Python program to count distinct elements in every sliding window of size k.
arr = [1, 2, 1, 3, 4, 2, 3]
k = 4

for i in range(len(arr) - k + 1):
    window = arr[i:i+k]
    distinct = set(window)

    print(len(distinct), end=" ")
```
