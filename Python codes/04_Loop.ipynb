```python
#14. Write a Python program to swap first and last digits of a number.

n=12345
last=n%10
temp=n
digit=0
while temp>0:
    digit+=1
    temp//=10

first=n//(10**(digit-1))
middle=(n%(10**(digit-1)))//10
result=last*(10**(digit-1))+(middle*10)+first

print("Number after swapping:", result)
```

```python
#18. Write a Python program to check whether a number is palindrome or not.
n=int(input("Enter number: "))
rev=0
temp=n
while temp>0:
    d=temp%10
    rev=rev*10+d
    temp//=10
    
if n==rev:
    print(n," is Palindrome")
else:
    print(n," is not Palindrome")
    
```

```python
# 23. Write a Python program to find HCF (GCD) of two numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b!=0:
    a,b=b,a%b
    
print("HCF =",a)

def GDC(a,b):
    if b==0:
        return a
    return GDC(b,a%b)
print("Function ans ",GDC(a,b))
```

```python
#26. Write a Python program to print all Prime numbers between 1 to n.

n = int(input("Enter n: "))

for num in range(2, n + 1):
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            break
    else:
        print(num, end=" ")
```

```python
#27. Write a Python program to find sum of all prime numbers between 1 to n.
n = int(input("Enter n: "))
ans=0
for num in range(2, n + 1):
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            break
    else:
        ans=ans+num
print("sum of prime number between 1 to",n,"is =",ans)
```

```python
#28. Write a Python program to check whether a number is Armstrong number or not.

n=int(input("Enter a num:"))
temp=n
n_len=len(str(n))
ans=0

while temp>0:
    d=temp%10
    ans=ans+d**n_len
    temp//=10
if ans == n:
    print(n, "is Armstrong Number")
else:
    print(n, "is Not Armstrong Number")
```

```python
#29. Write a Python program to print all Armstrong numbers between 1 to n.
n=int(input("Enter a num:"))
for i in range(1,n+1):
    temp=i
    n_len=len(str(i))
    ans=0

    while temp>0:
        d=temp%10
        ans=ans+d**n_len
        temp//=10
    if ans == i:
        print(i,end=" ")
```
