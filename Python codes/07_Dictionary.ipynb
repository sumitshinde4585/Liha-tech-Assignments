```python
# 12.  Write a Python program to sort a dictionary by values.

d= {"a": 30, "b": 10, "c": 20}
result=dict(sorted(d.items(),key=lambda x : x[1]))
print(result)
```

```python
d={"A": 75, "B": 92, "C": 88}
max_k=None
max_val=float("-inf")
for k,v in d.items():
    if v>max_val:
        max_val=v
        max_k=k
print("Key with maximum value:", max_k)
print("Maximum value:", max_val)

print(max(d, key=d.get))
```
